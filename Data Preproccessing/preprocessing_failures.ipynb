```python
import pandas as pd
import numpy as np
```

```python
failures = pd.read_csv("../Data/failures.csv")
```

```python
failures["datetime"] = pd.to_datetime(failures["datetime"])
```

```python
failures.head()
```

```python
failures = failures.sort_values(["machineID", "datetime"]).reset_index(drop=True)
```

```python
failure_dummies = pd.get_dummies(failures["failure"], prefix="failure")
failures = pd.concat([failures, failure_dummies], axis=1)
```

```python
failures.head()
```

```python
failures["days_since_last_failure"] = (
    failures.groupby("machineID")["datetime"]
    .transform(lambda x: x.diff().dt.total_seconds() / (3600 * 24))
)
#The machine that failed recently is not 
```

```python
failures["cumulative_failures"] = failures.groupby("machineID").cumcount() + 1
```
